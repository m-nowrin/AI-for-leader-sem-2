# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")

```

When plotting the number of visitors to a museum over two weeks, both of these methods have its own advantages and disadvantages, depending on the purpose of the analysis.

In the first approach, the two weeks are shown in side-by-side subplots. This makes it easier to focus on the individual pattern of each week without any visual interference from the other. This method helps in highlighting specific trends or anomalies unique to each week and prevents confusion that might occur if the data overlaps. However, the disadvantage is that direct comparison becomes more difficult, as the viewer must look back and forth between two separate graphs. If the axis scales are not kept consistent, the differences in visitor numbers might also appear exaggerated or minimized, which can lead to misinterpretation.

In the second approach, both weeks are plotted on the same set of axes. This allows for immediate visual comparison, making it easy to spot similarities and differences in visitor patterns across the two weeks. This method is especially useful if the goal is to compare overall trends or fluctuations day by day. However, if the lines overlap too much or if no proper legend is provided, it can become confusing for the viewer to distinguish which line represents which week. In your provided plots, using different colors helps solve this issue, but without labels or a legend, the interpretation can still be unclear.

The choice between these two methods depends on the goal of the presentation. If the focus is on understanding each week individually, separate subplots are more effective. If the focus is on comparing the two weeks directly, a combined plot is the better choice, provided it is clearly labeled.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Product names and years
products = ['Product A', 'Product B', 'Product C', 'Product D', 'Product E']
years = ['2008', '2009', '2010', '2011', '2012', '2013', '2014']

# Example price data (approximated from your chart)
prices = [
    [400, 420, 430, 400, 300, 270, 250],  # Product A
    [350, 400, 410, 380, 300, 270, 250],  # Product B
    [0, 0, 100, 180, 250, 230, 210],      # Product C (launched 2010)
    [0, 0, 0, 150, 250, 220, 210],        # Product D
    [0, 0, 0, 0, 0, 100, 210]            # Product E
]

# Convert to numpy for easy manipulation
prices = np.array(prices)

# Set position of bar groups
x = np.arange(len(products))
width = 0.1  # width of each bar

# Create plot
fig, ax = plt.subplots(figsize=(10,6))

# Plot each year's bar
for i in range(len(years)):
    ax.bar(x + i * width - 0.3, prices[:, i], width, label=years[i])

# Formatting the plot
ax.set_xlabel('Products')
ax.set_ylabel('Average Retail Price ($)')
ax.set_title('Average Retail Product Price Per Year')
ax.set_xticks(x)
ax.set_xticklabels(products)
ax.set_ylim(0, 500)
ax.legend(title='Year')
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
